# ============================
# 1. Setup & Load Dependencies
# ============================

```python
# Install necessary library (if not already installed)
#!pip install seaborn --quiet

import pandas as pd
import numpy as np
import seaborn as sns
import seaborn as sns
import matplotlib.pyplot as plt


# Make plots display nicely
sns.set()
```

# ============================
# 2. Load a Sample Dataset
# ============================

```python
# Using the built-in 'tips' dataset from seaborn
df = sns.load_dataset('tips')


# Display first few rows
df.head()


```

# ============================
# 3. Inspect Data Structure
# ============================

```python
print("\n🔹 Data Shape (rows, columns):")
print(df.shape)


print("\n🔹 Column Names:")
print(df.columns.tolist())


print("\n🔹 Data Types:")
print(df.dtypes)


print("\n🔹 Check Missing Values:")
print(df.isnull().sum())


```

# ============================
# 4. Basic Summary Statistics
# ============================

```python
print("\n📊 Summary Statistics (numerical):")
display(df.describe())

cat_cols = df.select_dtypes(include=['object', 'category']).columns

print("\n📊 Summary Statistics (categorical):")


if len(cat_cols) > 0:
 display(df[cat_cols].describe())
else:
 print("No categorical columns found in this dataset.")



```

# ============================
# 5. Basic Visualizations
# ============================



```python

# 5.1 Distribution of total bill
plt.figure(figsize=(6,4))
sns.histplot(df['total_bill'], kde=True, bins=20)
plt.title("Distribution of Total Bill")
plt.show()
print("\n")

# 5.2 Total Bill vs Tip Scatter Plot
plt.figure(figsize=(6,4))
sns.scatterplot(data=df, x='total_bill', y='tip')
plt.title("Total Bill vs Tip")
plt.show()
print("\n")

# 5.3 Boxplot: Tips by Day
plt.figure(figsize=(6,4))
sns.boxplot(data=df, x='day', y='tip')
plt.title("Tip Amount by Day of Week")
plt.show()
print("\n")

# 5.4 Correlation Heatmap
plt.figure(figsize=(6,4))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
