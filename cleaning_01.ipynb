#Simple cleaning data

```python
import pandas as pd
import numpy as np
```

# Covers:
# - Loading data
# - Checking missing values
# - Filling missing values
# - Dropping duplicates
# - Fixing inconsistent text

# -----------------------------------
# 1. Create a small messy dataset
# -----------------------------------

```python
data = {
"name": ["Alice", "Bob", "alice", "Charlie", None],
"age": [25, None, 22, 30, 25],
"city": ["New York", "new york", "London", None, "London"],
"score": [85, 90, None, 70, 85]
}
```

```python
df = pd.DataFrame(data)
print("Raw Data:\n", df, "\n")
```

# -----------------------------------
# 2. Check missing values
# -----------------------------------

```python
print("Missing Values:\n", df.isna().sum(), "\n")
```

# -----------------------------------
# 3. Fill missing values
# -----------------------------------

```python
df['age'] = df['age'].fillna(df['age'].median()) # numeric fill
df['city'] = df['city'].fillna("Unknown") # text fill
df['score'] = df['score'].fillna(df['score'].mean()) # numeric fill
df['name'] = df['name'].fillna("Unknown") # text fill

print("After Filling Missing Values:\n", df, "\n")
```

# -----------------------------------
# 4. Remove duplicates
# -----------------------------------

```python
df = df.drop_duplicates()
print("After Removing Duplicates:\n", df, "\n")
```

# -----------------------------------
# 5. Fix inconsistent text (lowercase + title case)
# -----------------------------------

```python
df['name'] = df['name'].str.title()
df['city'] = df['city'].str.title()


print("After Cleaning Text Columns:\n", df, "\n")
```
